# 02 · Operators

**Topics covered:** arithmetic · relational · logical · assignment · bitwise · identity · membership · precedence

## 1. Arithmetic operators  `+  -  *  /  //  %  **`

```python
a, b = 17, 5
print("a + b  =", a + b)
print("a - b  =", a - b)
print("a * b  =", a * b)
print("a / b  =", a / b)     # true division -> float
print("a // b =", a // b)    # floor division
print("a % b  =", a % b)     # remainder
print("a ** b =", a ** b)    # power
```

## 2. Relational (comparison) operators  `<  <=  >  >=  ==  !=`
Always give `True` / `False`.

```python
x, y = 10, 20
print(x < y, x <= y, x > y, x >= y, x == y, x != y)
```

## 3. Logical operators  `and  or  not`

```python
age = 25
has_id = True
print(age >= 18 and has_id)    # both must be True
print(age < 18 or has_id)      # at least one True
print(not has_id)              # reverses

# short-circuit behaviour
print(0 or "default")          # first truthy value
print(5 and 10)                # last value if all truthy
```

## 4. Assignment operators  `=  +=  -=  *=  /=  //=  %=  **=`

```python
n = 10
n += 5;  print(n)    # 15
n -= 3;  print(n)    # 12
n *= 2;  print(n)    # 24
n /= 4;  print(n)    # 6.0
n //= 4; print(n)    # 1.0
n %= 1;  print(n)    # 0.0

p = 2
p **= 5
print(p)             # 32
```

## 5. Bitwise operators  `&  |  ^  ~  <<  >>`
They work on the **binary representation** of integers.

```python
a, b = 12, 10          # 1100 and 1010
print(a & b, bin(a & b))   # AND  -> 1000
print(a | b, bin(a | b))   # OR   -> 1110
print(a ^ b, bin(a ^ b))   # XOR  -> 0110
print(~a)                  # NOT  -> -(a+1)
print(a << 1)              # left shift  = a * 2
print(a >> 1)              # right shift = a // 2
```

## 6. Identity operators  `is   is not`
Check whether two names refer to the **same object** in memory (not just equal value).

```python
l1 = [1, 2, 3]
l2 = [1, 2, 3]
l3 = l1
print(l1 == l2)       # True  -> same value
print(l1 is l2)       # False -> different objects
print(l1 is l3)       # True  -> same object
print(l1 is not l2)   # True

x = None
print(x is None)      # always use `is` to compare with None
```

## 7. Membership operators  `in   not in`

```python
fruits = ["apple", "banana", "cherry"]
print("apple" in fruits)
print("mango" not in fruits)
print("a" in "banana")
print(2 in {1: "x", 2: "y"})   # checks dictionary KEYS
```

## 8. Operator precedence (high → low)
`()` → `**` → `+x -x ~x` → `* / // %` → `+ -` → `<< >>` → `&` → `^` → `|` → comparisons / `is` / `in` → `not` → `and` → `or`

```python
print(2 + 3 * 4)        # 14
print((2 + 3) * 4)      # 20
print(2 ** 3 ** 2)      # 512  (** is right-associative)
print(not True or True) # True -> (not True) or True
```
