# 17 · OOP – Abstraction

Hide the **implementation details** and expose only what the user needs to know ("what", not "how").
Example: you press the brake pedal – you don't need to know the hydraulics.

**Topics covered:** `abc` module · abstract class · abstract method · concrete methods · rules · practical examples

## 1. How Python does it
- `from abc import ABC, abstractmethod`
- A class that inherits `ABC` and has at least one `@abstractmethod` is an **abstract class**
- You **cannot create an object** of an abstract class
- Every child class **must implement** all abstract methods

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

try:
    s = Shape()
except TypeError as e:
    print("TypeError:", e)
```

## 2. Implementing the abstract class

```python
class Rectangle(Shape):
    def __init__(self, l, b):
        self.l, self.b = l, b
    def area(self):
        return self.l * self.b

class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):
        return round(3.14159 * self.r ** 2, 2)

print(Rectangle(4, 5).area())
print(Circle(3).area())
```

## 3. Child must implement ALL abstract methods

```python
class Incomplete(Shape):
    pass          # forgot to implement area()

try:
    Incomplete()
except TypeError as e:
    print("TypeError:", e)
```

## 4. Abstract + concrete methods together

```python
class Vehicle(ABC):
    def start(self):                 # concrete – shared by all children
        print("Engine started")

    @abstractmethod
    def wheels(self):                # abstract – each child must define
        pass

class Car(Vehicle):
    def wheels(self):
        return 4

class Bike(Vehicle):
    def wheels(self):
        return 2

for v in (Car(), Bike()):
    v.start()
    print(type(v).__name__, "has", v.wheels(), "wheels")
```

## 5. Real-life example – payment system

```python
class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class UPI(Payment):
    def pay(self, amount):
        print(f"Paid {amount} using UPI")

class Card(Payment):
    def pay(self, amount):
        print(f"Paid {amount} using Card")

def checkout(method: Payment, amount):
    method.pay(amount)          # caller doesn't care HOW it is paid

checkout(UPI(), 500)
checkout(Card(), 1200)
```

## 6. Abstract class with constructor
Abstract classes can have `__init__`, normal attributes and even abstract properties.

```python
class Employee(ABC):
    def __init__(self, name):
        self.name = name

    @abstractmethod
    def salary(self):
        pass

    def show(self):
        print(self.name, "->", self.salary())

class FullTime(Employee):
    def salary(self):
        return 50000

class Intern(Employee):
    def salary(self):
        return 10000

FullTime("Rahul").show()
Intern("Shubham").show()
```
