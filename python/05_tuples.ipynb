# 05 · Tuples

**Topics covered:** properties · creating · accessing · concatenation · slicing · deleting · methods · built-in functions · conversion · unpacking · practice programs

## 1. Properties
- **Immutable** – cannot be changed after creation
- **Ordered**, allows duplicates
- Can hold different data types
- Created with `()` and comma separated items
- Operations: create, access, concatenate, slice, delete

```python
t = (10, 'hii', 10)
print(t)
print(t[0])                # accessing
t2 = (45, 8.9)
t3 = t + t2                # concatenation
print(t3)
print(t3[0:2])             # slicing
del t3                     # delete the whole tuple (single items cannot be deleted)

# one-element tuple needs a trailing comma
single = (5,)
print(type(single), type((5)))

# parentheses are optional (tuple packing)
t = 1, 2, 3
print(type(t))
```

## 2. Methods & built-in functions

```python
t = (10, 'hii', 10)
t2 = (45, 8.9)
print(t.index(10))       # index of first occurrence
print(t.index(10, 1))    # search starting from index 1
print(t.count(10))       # how many times

t1 = ()
print(type(t1))
print(all(t1))           # True for an empty tuple
print(any(t1))           # False for an empty tuple
print(len(t2), max(t2), min(t2), sum(t2))
print(sorted(t2))        # returns a LIST
```

## 3. Conversion between types

```python
t4 = [1, 4, 6, "t"]
print(tuple(t4))      # list  -> tuple
print(list(t4))       # tuple -> list
print(set(t4))        # -> set
print(str(t4))        # converts whole thing to text like "[1, 4, 6, 't']"

w = ['a', 'b', 'c']
print("".join(w))     # join characters into a string (items must be str)
```

## 4. Practice programs

```python
# Q1 create tuple apple, banana, cherry
fruits = ("apple", "banana", "cherry")
print(fruits[0])           # first element
print(fruits[2])           # third element

# Q2 length of tuple
print(len(fruits))

# Q3 check if "banana" is present
if "banana" in fruits:
    print("is present")
else:
    print("not present")
print("banana" in fruits)
```

```python
# Unpack a tuple into variables
person = ("John", 25, "Engineer")
name, age, profession = person
print(name)
print(age)
print(profession)

# extended unpacking
first, *middle, last = (1, 2, 3, 4, 5)
print(first, middle, last)
```

```python
# Maximum and minimum
t1 = (12, 3, 4)
print(min(t1))
print(max(t1))

# repeat a number 10 times  (don't forget the comma!)
t1 = (5,) * 10
print(t1)

# tuple of pairs -> dictionary
stud = (("name", "shubham"), ("age", 22), ("Course", "AgenticAi"))
d1 = dict(stud)
print(d1)
```

## 5. Assignments

```python
# 1. concatenate two tuples
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)
print(tuple1 + tuple2)

# 2. slice second to fourth elements
numbers = (10, 20, 30, 40, 50)
print(numbers[1:4])

# 3. count occurrences of 2
numbers = (1, 2, 2, 3, 2, 4)
print(numbers.count(2))

# 4. "remove" an item by creating a new tuple
t1 = (10, 20, 30, 40)
t2 = tuple(x for x in t1 if x != 30)       # way 1: filter
t3 = t1[:2] + t1[3:]                       # way 2: slicing
print(t2, t3)

# 5. length of the longest string in a tuple of strings
mytuple = ("abc", "whftg", "ashrrbf")
print(max(len(t) for t in mytuple))
print(len(max(mytuple, key=len)))
print(max(mytuple, key=len))
```

## 6. Why use tuples?
Faster than lists, safe from accidental change, can be **dictionary keys**, ideal for fixed data like coordinates `(x, y)`.
