## 1. What is a function?
A reusable block of code that runs only when called. Benefits: reuse, readability, easy debugging.

```python
def function_name(parameters):
    # body
    return value      # optional
```

````python
def greet():
    print("Hello World!!")

greet()
greet()
````

## 2. Parameters vs Arguments
- **Parameter** = variable in the function definition
- **Argument** = actual value passed when calling

````python
def FN(N, S):               # N, S are parameters
    fn = N + " " + S
    return fn

print(FN(N="shubham", S="jadhav"))   # keyword arguments
print(FN("shubham", "jadhav"))       # positional arguments
````

## 3. Return value

````python
def factorial(num):
    fact = 1
    i = 1
    while i <= num:
        fact *= i
        i += 1
    return fact

n_ = 6
m = 7
print(factorial(n_))
print(factorial(m))

# returning multiple values (comes back as a tuple)
def min_max_sum(a, b, c):
    nums = (a, b, c)
    return min(nums), max(nums), sum(nums)

lo, hi, tot = min_max_sum(4, 9, 2)
print(lo, hi, tot)
````

## 4. Types of arguments

### a) Positional arguments – order matters

````python
def stud(nm, age):
    print(nm, age)

stud("shubham", 22)
````

### b) Keyword arguments – order does not matter

````python
def emp(nm, sal):
    print(nm, sal)

emp(sal=50000, nm="shubham")
````

### c) Default arguments – used when no value is passed

````python
def cntry(nm="ind"):
    print(nm)

cntry()
cntry("USA")

# your Book example – FIXED: the original had no defaults, so Book() raised TypeError
def Book(title="unknown", author="unknown"):
    return f"The title is {title} and author is {author}"

print(Book("agnipath", "a.p.j.kalam"))
print(Book())
````

> ⚠️ Default arguments must come **after** non-default ones: `def f(a, b=2)` ✔  `def f(a=1, b)` ✘

### d) Variable length positional arguments – `*args` (collected as a tuple)

````python
def total(*num):
    print(num, "->", sum(num))

total(10, 20, 30)
total(1, 2)

def Add(*args):
    s = 0
    for i in args:
        s = s + i
    return s

print(Add(12, 23, 34, 45, 456, 7))
print(Add(12, 23, 34, 45))
````

### e) Variable length keyword arguments – `**kwargs` (collected as a dict)

````python
def details(**data):
    print(data)

details(nm="shubham", city="pune")

def test(**kwargs):
    for key, value in kwargs.items():
        print(key, value)

test(Subject="Python", Marks=78)
````

### f) Combining everything
Order: `normal → *args → default/keyword → **kwargs`

````python
def demo(*args, **kwargs):
    for i in args:
        print(i)
    for key, value in kwargs.items():
        print(key, value)

demo(12, 23, 24, name="ABC")

def mix(a, b=10, *args, **kwargs):
    print(a, b, args, kwargs)

mix(1)
mix(1, 2, 3, 4, x=5)
````

### g) Unpacking a list / dict while calling

````python
def sum_all(*args):
    return sum(args)

l1 = [1, 2, 3, 4]
print(sum_all(*l1))            # * unpacks the list into separate arguments

def intro(name, age):
    print(name, age)

d = {"name": "sj", "age": 22}
intro(**d)                     # ** unpacks the dict into keyword arguments
````

## 5. Scope – local vs global

````python
x = 10                      # global

def show():
    x = 5                   # local (different variable)
    print("inside:", x)

show()
print("outside:", x)

def change():
    global x
    x = 99

change()
print(x)
````

## 6. `lambda` (anonymous) functions + `map`, `filter`

````python
square = lambda x: x * x
add = lambda a, b: a + b
print(square(5), add(2, 3))

num = [1, 2, 3, 4]
print(list(map(lambda x: x * 2, num)))
print(list(filter(lambda x: x % 2 == 0, num)))
print(sorted(["pear", "fig", "banana"], key=lambda s: len(s)))
````

## 7. Recursion – a function calling itself

````python
def fact(n):
    if n <= 1:          # base case
        return 1
    return n * fact(n - 1)

print(fact(5))

def fib(n):
    return n if n < 2 else fib(n - 1) + fib(n - 2)

print([fib(i) for i in range(10)])
````
