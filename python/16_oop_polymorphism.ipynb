# 16 · OOP – Polymorphism

*Poly = many, morph = forms* → the **same name / interface behaves differently** in different situations.

**Topics covered:** method overriding · method overloading (default / `*args`) · duck typing · operator overloading · built-in polymorphism

## 1. Method overriding (run-time polymorphism)

```python
class animal:
    def sound(self):
        return "Animal makes a sound."

class dog(animal):
    def sound(self):
        return "Dog barks."

class cat(animal):
    def sound(self):
        return "Cat meows."

# same call -> different behaviour
for a in (animal(), dog(), cat()):
    print(a.sound())
```

## 2. Method overloading
Python does **not** support true overloading (a later `def` with the same name replaces the earlier one). We imitate it with **default arguments** or `*args`.

```python
# default argument
class Calculator:
    def add(self, a, b, c=0):
        return a + b + c

calc = Calculator()
print(calc.add(5, 10))        # 15
print(calc.add(5, 10, 15))    # 30
```

```python
# *args – any number of arguments
class Calc:
    def add(self, *args):
        return sum(args)

c = Calc()
print(c.add(1, 2))
print(c.add(1, 2, 3, 4, 5))

# the second def REPLACES the first
class Demo:
    def show(self, a):
        print("one arg")
    def show(self, a, b):
        print("two args")

try:
    Demo().show(1)
except TypeError as e:
    print("TypeError:", e)
```

## 3. Duck typing
*"If it walks like a duck and quacks like a duck, it is a duck."* – no inheritance needed; only the method name matters.

```python
class Dog:
    def speak(self): return "Woof"
class Cat:
    def speak(self): return "Meow"
class Robot:
    def speak(self): return "Beep"

def make_it_speak(thing):
    print(thing.speak())

for obj in (Dog(), Cat(), Robot()):
    make_it_speak(obj)
```

## 4. Polymorphism with a parent class (shapes)

```python
class Shape:
    def area(self):
        return 0

class Rectangle(Shape):
    def __init__(self, l, b):
        self.l, self.b = l, b
    def area(self):
        return self.l * self.b

class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):
        return round(3.14159 * self.r ** 2, 2)

shapes = [Rectangle(4, 5), Circle(3), Shape()]
for s in shapes:
    print(type(s).__name__, "area =", s.area())
```

## 5. Operator overloading (magic methods)

```python
class Vector:
    def __init__(self, x, y):
        self.x, self.y = x, y
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)
    def __eq__(self, other):
        return self.x == other.x and self.y == other.y
    def __str__(self):
        return f"Vector({self.x}, {self.y})"

v1, v2 = Vector(1, 2), Vector(3, 4)
print(v1 + v2)
print(v1 == Vector(1, 2))
```

## 6. Built-in polymorphism

```python
print(len("python"), len([1, 2, 3]), len({"a": 1}))   # len() works on many types
print(1 + 2, "a" + "b", [1] + [2])                    # + works differently by type
print(2 * 3, "ab" * 3, [0] * 3)
```
