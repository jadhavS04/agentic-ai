**Topics covered:** calling vs referencing a function · iterable vs iterator · `iter()` / `next()` · comprehensions (list / set / dict) · generator functions with `yield` · generator expressions · practice programs

## 1. Function call vs function reference

```python
def test():
    print("Function called!!")

a = test()     # CALLS test(); a stores its return value (None)
print(a)

a = test       # no brackets -> a is just another name for the function
a()            # now call it
```

## 2. Iterable vs Iterator
- **Iterable** – can be looped over (list, tuple, str, dict, set, range)
- **Iterator** – object that returns the **next value one at a time** using `next()`; created from an iterable with `iter()`

```python
list1 = [123, 345, 7, 65857, 98, 456, 63, 4]
g = iter(list1)          # iterable -> iterator
print(next(g))
print(next(g))
print(next(g))

# when exhausted -> StopIteration
small = iter([1, 2])
print(next(small), next(small))
try:
    next(small)
except StopIteration:
    print("no more items (StopIteration)")
```

## 3. Comprehensions (compact loops)

```python
l1 = [1, 23, 34, 56, 67, 89]
print([i * 2 for i in l1])               # list comprehension
print({i * 2 for i in l1})               # set comprehension  { }
print([i for i in l1 if i % 2 == 0])     # with a condition
print({i: i ** 2 for i in l1[:3]})       # dict comprehension
```

## 4. Generators
A **generator** produces a *sequence of values, one at a time*, using `yield`.
- Defined like a normal function but uses `yield` instead of `return`
- Calling it does **not** run the body – it returns a generator object
- Each `next()` resumes from where it last paused
- **Memory-efficient** – values are produced lazily

```python
def demo():
    yield 10
    yield 20
    yield 30

d = demo()
print(d)            # generator object
print(next(d))      # 10
print(next(d))      # 20
print(next(d))      # 30
```

```python
# generators are used naturally in for loops
for value in demo():
    print(value)

print(list(demo()))
```

```python
# yield pauses, return stops
def show_steps():
    print("step 1")
    yield 1
    print("step 2")
    yield 2
    print("end")

g = show_steps()
print(next(g))
print(next(g))
try:
    next(g)
except StopIteration:
    print("finished")
```

## 5. Generator over a list

```python
def genFun(lst1):
    for i in lst1:
        yield i

list1 = [123, 345, 7, 65857, 98, 456, 63, 4]
g = genFun(list1)
print(next(g))
print(next(g))
print(list(g))      # remaining values
```

## 6. Generator expression
Change the brackets of a list comprehension from `[]` to `()` → it becomes a generator (iterator).

```python
list1 = [123, 345, 7, 65857, 98, 456, 63, 4]
g = (i * 2 for i in list1)
print(type(g))
print(next(g))
print(next(g))

# memory comparison
import sys
print(sys.getsizeof([x for x in range(100000)]), "bytes (list)")
print(sys.getsizeof((x for x in range(100000))), "bytes (generator)")
```

## 7. Practice programs

```python
# Gen_Seq(): numbers from 101 to 200
# (your original used range(101, 202) which also produced 201 – stop value is excluded)
def Gen_Seq():
    for i in range(101, 201):
        yield i

g = Gen_Seq()
print(next(g), next(g), next(g))
nums = list(Gen_Seq())
print(nums[0], nums[-1], len(nums))
```

```python
# generator with start and end using a while loop
start = int(input("enter start = "))
end = int(input("enter end = "))

def Gen_Seq(start, end):
    i = start
    while i < end:
        yield i
        i += 1

g = Gen_Seq(start, end)
print(next(g))
print(next(g))
print(list(g))
```

```python
# Fibonacci generator
def fib(limit):
    a, b = 0, 1
    while a <= limit:
        yield a
        a, b = b, a + b

print(list(fib(100)))

# even numbers generator
def evens(n):
    for i in range(n + 1):
        if i % 2 == 0:
            yield i

print(list(evens(20)))
```
