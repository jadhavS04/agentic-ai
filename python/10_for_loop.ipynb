**Topics covered:** `range()` · looping over list / string / dict · `enumerate` · `zip` · nested loops · practice problems

## 1. `range(start, stop, step)`
`stop` is excluded; defaults: `start=0`, `step=1`.

```python
for i in range(6):          # 0..5
    print(i, end=" ")
print()
for i in range(1, 11):      # 1..10
    print(i, end=" ")
print()
for i in range(1, 11, 2):   # odd 1..9
    print(i, end=" ")
print()
for i in range(10, 0, -2):  # counting down
    print(i, end=" ")
```

## 2. Looping over sequences

```python
Names = ["Vg", "jg", "fg", "gt"]
print(len(Names))
for i in Names:
    print(i)

str1 = "hef"
for c in str1:
    print(c)

# first letters / upper-case names
r = []
for i in Names:
    r.append(i[0])
print(r)

r = []
for i in Names:
    r.append(i.upper())
print(r)
```

```python
# index + value together
for idx, name in enumerate(Names):
    print(idx, name)

# two lists together
ages = [20, 21, 22, 23]
for n_, a_ in zip(Names, ages):
    print(n_, a_)

# looping a dictionary
dict1 = {"k1": 1, "k2": 2, "k3": 3}
s = 0
for i in dict1.values():
    s += i
print(s)
```

## 3. Problems

```python
# factorial using for
N = int(input("Enter num: "))
fact = 1
for i in range(1, N + 1):
    fact *= i
print(f"factorial of {N} is {fact}")
```

```python
# split a list into even and odd
l1 = [1, 2, 3, 4, 5, 6]
l2even = []
l3odd = []
for i in l1:
    if i % 2 == 0:
        l2even.append(i)
    else:
        l3odd.append(i)
print(l2even, l3odd)
```

```python
# vowels and consonants
l1 = ['a', 'e', 'i', 'o', 'u', 'c', 'f', 't', 'h', 'w']
vowels = []
consonants = []
for i in l1:
    if i in ('a', 'e', 'i', 'o', 'u'):
        vowels.append(i)
    else:
        consonants.append(i)
print(vowels, consonants)
```

```python
# multiplication table of N
N = int(input("Enter Number: "))
for i in range(1, 11):
    print(N, "x", i, "=", N * i)
```

```python
# count a character in a string
str1 = input("enter string: ")
char = input("Enter char: ")
c = 0
for i in str1:
    if i == char:
        c += 1
print(c)
```

```python
# count vowels
str1 = input("enter: ")
vowels = ("a", "e", "i", "o", "u", "A", "E", "I", "O", "U")
c = 0
for i in str1:
    if i in vowels:
        c += 1
print(c)
```

```python
# sum of odd numbers 1..100 (two ways)
s = 0
for i in range(1, 101):
    if i % 2 != 0:
        s += i
print(s)

s = 0
for i in range(1, 101, 2):
    s += i
print(s)
```

```python
# sum of list
l1 = [1, 2, 3]
c = 0
for i in l1:
    c = i + c
print(c)

# sum only numbers from a mixed list (type() vs isinstance())
l1 = [1, "a", 2.4, "b"]
c = 0
for i in l1:
    if type(i) == int or type(i) == float:
        c += i
print(c)

c = 0
for i in l1:
    if isinstance(i, (int, float)):
        c += i
print(c)
```

## 4. Nested loops & patterns

```python
# common elements of two lists
l1 = [1, 2, 3, 4]
l2 = [3, 4, 5]
for i in l1:
    for j in l2:
        if i == j:
            print(i)

# star pattern
for i in range(1, 6):
    print("*" * i)

# number triangle
for i in range(1, 5):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
```

```python
# palindromes from a list
l1 = ["abc", "abccba", "madam"]
for i in l1:
    if i == i[::-1]:
        print(i)
```

## 5. `for … else`, `break`, `continue`

```python
for i in range(1, 8):
    if i == 3:
        continue       # skip 3
    if i == 6:
        break          # stop at 6
    print(i, end=" ")
print()

# for-else: else runs only when no break happened
for i in range(2, 10):
    if 25 % i == 0:
        print("found divisor", i)
        break
else:
    print("no divisor")
```
