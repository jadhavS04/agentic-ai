## 1. File modes

| Mode | Meaning |
|---|---|
| `r` | read (default) – error if file missing |
| `w` | write – **creates / overwrites** |
| `a` | append – adds at the end |
| `x` | create – error if exists |
| `r+` | read + write |
| `w+` | write + read (overwrites) |
| `b` | binary (e.g. `rb`, `wb`) |

## 2. Write → read → append

```python
file = open("hii.txt", "w")      # create / overwrite
file.write("world")
file.close()

file = open("hii.txt", "r")
data = file.read()
print(data)
file.close()

file = open("hii.txt", "a")      # append keeps old data
file.write("\nworld")
file.write("world")
file.close()

print(open("hii.txt").read())
```

## 3. `read()`, `readline()`, `readlines()`

```python
file = open("stud.txt", "w")
file.write("Hello, World!\nhi\nhi")
file.close()

file = open("stud.txt", "r")
print(repr(file.readline(10)))     # first 10 characters of the line
file.close()

file = open("stud.txt", "r")
e = file.readlines()               # list of lines (with \n)
print(e)
print(len(e))
file.close()

# iterate line by line (memory friendly)
for line in open("stud.txt"):
    print(line.strip())
```

## 4. `with` statement – closes the file automatically

```python
id_ = input("enter id: ")
nm = input("enter name: ")
with open("emp.txt", "w") as file:
    file.write("id=" + id_ + "\n")
    file.write("nm=" + nm + "\n")
# no file.close() needed

with open("emp.txt") as f:
    print(f.read())
```

```python
name = input("enter name: ")
passw = input("enter password: ")
with open("emp.txt", "w") as file:
    file.write("name=" + name + "\n")
    file.write("password=" + passw + "\n")

with open("emp.txt", "r") as file:
    print(file.read())
```

## 5. Practice – student details

```python
name = input("enter name: ")
age = input("enter age: ")
course = input("enter course: ")
city = input("enter city: ")
with open("student.txt", "w") as file:
    file.write("name=" + name + "\n")
    file.write("age=" + age + "\n")
    file.write("course=" + course + "\n")
    file.write("city=" + city + "\n")
print("data saved successfully")

with open("student.txt", "r") as file:
    print(file.read())
```

```python
# college details, then print each line separately
with open("college.txt", "w") as file:
    file.write("collegename=" + input("enter collegename: ") + "\n")
    file.write("department=" + input("enter department: ") + "\n")
    file.write("course=" + input("enter course: ") + "\n")
    file.write("academic_year=" + input("enter academic year: ") + "\n")

with open("college.txt", "r") as file:
    data = file.readlines()
print(data)
for line in data:
    print(line.strip())
```

## 6. Append data without deleting old data

```python
# first student already in the file
with open("students.txt", "w") as f:
    f.write("Name: sj\nCourse: ai\nMarks: 89\n")

# accept another student and APPEND
with open("students.txt", "a") as f:
    f.write("Name: " + input("Enter student's name: ") + "\n")
    f.write("Course: " + input("Enter student's course: ") + "\n")
    f.write("Marks: " + input("Enter student's marks: ") + "\n")

with open("students.txt") as f:
    for line in f.readlines():
        print(line.strip())
```

## 7. Count lines / students

```python
with open("students.txt") as f:
    data = f.readlines()

print("Total lines:", len(data))
print("Total number of students:", len(data) // 3)   # 3 lines per student
# (your original printed len(data) = 6 lines, which is lines, not students)
```

## 8. `seek()` and `tell()` – moving the file pointer

```python
with open("abc.txt", "w+") as file:
    file.write("Welcome to exception handling\nwelcome to exception logging ")

with open("abc.txt", "r") as file:
    file.seek(15)            # jump to byte 15
    print(file.tell())       # current position
    print(file.read())       # reads from there
    file.seek(0)
    print(file.read(7))      # first 7 characters
```
