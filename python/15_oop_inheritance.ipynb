## 1. What is inheritance?
A child (derived) class **acquires the attributes and methods** of a parent (base) class → code reuse.

```
Single:       A → B
Multilevel:   A → B → C
Multiple:     A, B → C
Hierarchical: A → B, A → C
Hybrid:       combination of the above
```

## 2. Single inheritance

````python
class animal:
    def sound(self):
        return "Animal makes a sound."

class dog(animal):
    def sound(self):
        return "Dog barks."

class cat(animal):
    def sound(self):
        return "Cat meows."

a = animal()
d = dog()
c = cat()
print(a.sound())
print(d.sound())
print(c.sound())
````

````python
class Person:
    def __init__(self, name):
        self.name = name
    def show(self):
        print("Name:", self.name)

class Student(Person):
    def __init__(self, name, roll):
        super().__init__(name)      # call parent constructor
        self.roll = roll
    def show_all(self):
        self.show()                 # inherited method
        print("Roll:", self.roll)

s = Student("Shubham", 20)
s.show_all()
````

## 3. Multilevel inheritance

````python
class Grandparent:
    def show_grandparent(self):
        print("This is the grandparent class method.")

class Parent(Grandparent):
    def show_parent(self):
        print("This is the parent class method.")
        self.show_grandparent()

class Child(Parent):
    def show_child(self):
        print("This is the child class method.")
        self.show_parent()

c = Child()
c.show_child()
````

## 4. Multiple inheritance

````python
class Father:
    def show_father(self):
        print("Father has a car")

class Mother:
    def show_Mother(self):
        print("Mother has a house")

class Child(Father, Mother):
    def show_Child(self):
        print("child has bike")

c = Child()
c.show_father()
c.show_Mother()
c.show_Child()
````

````python
class Employee:
    def emp_sal(self):
        self.salary = 10000

class bonus:
    def emp_bonus(self):
        self.bonus = 5000

class Total(Employee, bonus):
    def total_sal(self):
        self.emp_sal()
        self.emp_bonus()
        total = self.salary + self.bonus
        print("Total salary =", total)

Total().total_sal()
````

````python
class product:
    def product_details(self):
        self.productnm = "Laptop"
        self.price = 45000

class payment:
    def payment_details(self):
        self.payment_mode = "UPI"

class order(product, payment):
    def place_order(self):
        self.product_details()
        self.payment_details()
        print("Product Details:", self.productnm)
        print("Price:", self.price)
        print("Payment Mode:", self.payment_mode)

order().place_order()
````

## 5. Hierarchical & hybrid inheritance

````python
# Hierarchical: one parent, many children
class Vehicle:
    def start(self):
        print("Vehicle started")

class Car(Vehicle):
    pass

class Bike(Vehicle):
    pass

Car().start()
Bike().start()

# Hybrid: hierarchical + multiple
class A:
    def who(self): print("A")
class B(A): pass
class C(A): pass
class D(B, C): pass
D().who()
````

## 6. Method Resolution Order (MRO)
Python searches the class itself → then left-to-right through the parents.

````python
class A:
    def show(self): print("A")
class B(A):
    def show(self): print("B")
class C(A):
    def show(self): print("C")
class D(B, C):
    pass

D().show()          # B  (first parent wins)
print(D.__mro__)
print([c.__name__ for c in D.mro()])
````

## 7. Method overriding & `super()`
Child redefines a parent method. `super()` still lets you call the parent version.

````python
class Parent:
    def show(self):
        print("This is the parent class method.")

class Child(Parent):
    def show(self):
        print("This is the child class method.")
        super().show()

Child().show()
````

## 8. `isinstance`, `issubclass`

````python
class A: pass
class B(A): pass

b = B()
print(isinstance(b, B), isinstance(b, A), isinstance(b, int))
print(issubclass(B, A), issubclass(A, B))
````
