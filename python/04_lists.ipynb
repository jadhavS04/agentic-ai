# 04 · Lists

**Topics covered:** list properties · creating · accessing · adding · removing · sorting · nested lists · `map` / `filter` / `lambda` · list comprehension · practice programs

## 1. Properties
- Stores multiple items in a single variable
- **Duplicates allowed**, **mixed types** allowed
- **Mutable** (changeable) and **ordered**
- Access by position (index) → `my_list[0]`
- Most versatile data type in Python

```python
my_list = [10, "hello", 3.14, True, 20]
print(my_list)

# Q1 list with mixed values
mylist = [1, 2, 3, 4, 'h', 'a', 'g', 'ee', 'rr', "tt"]
print(mylist)
```

## 2. Creating & accessing

```python
a = [1, 2, 3, 4, 5]
b = list((1, 2, 3, 4, 5))      # list() constructor
print(a, b)
print(a[0], a[-1], a[1:4], a[::-1])
```

## 3. Adding elements

```python
a = [1, 2, 3, 4, 5]
a.append(60)            # add at the end
print(a)
a.insert(0, 90)         # add at a specific position
print(a)
a.extend([34, 56, 78])  # add multiple elements
print(a)
a[1] = 25               # change an element
print(a)
```

## 4. Removing elements

```python
a = [1, 2, 3, 4, 5]
a.remove(4)      # removes the first OCCURRENCE of the value
print(a)
a.pop()          # removes the LAST element
print(a)
a.pop(1)         # removes the element at index 1
print(a)
del a[1]         # delete by index
print(a)

b = [1, 2, 3]
b.clear()        # empties the list
print(b)
```

## 5. Sorting, searching & other methods

```python
nums = [311, 2, 3, 4, 5, 6]
nums.sort()                      # in place
print(nums)
nums.sort(reverse=True)
print(nums)
print(sorted(nums))              # returns a NEW list
nums.reverse()
print(nums)

data = [5, 3, 5, 1]
print(data.count(5), data.index(3))
print(len(data), min(data), max(data), sum(data))
copy1 = data.copy()              # shallow copy
print(copy1)
```

## 6. Nested lists

```python
list2 = [[1, 2, 3, 4, 5],
         ["a", "b", "c"],
         [1.2, 4.5]]
print(list2[1][1])      # 'b'
for row in list2:
    print(row)
```

## 7. `map`, `filter`, `lambda`

```python
list1 = [2, 3, 4, 5, 6, 311]
print(list(map(lambda x: x * 2, list1)))          # transform every item
print(list(filter(lambda x: x % 2 == 0, list1)))  # keep only even items
```

## 8. List comprehension
`[expression for item in iterable if condition]`

```python
l1 = [1, 23, 34, 56, 67, 89]
print([i * 2 for i in l1])
print([i for i in l1 if i % 2 == 0])
print({i * 2 for i in l1})          # set comprehension
```

## 9. Practice programs

```python
# Q2 favourite fruits list
Myfavfrt = ['mango', 'banana', "apple", "grapes"]
print(Myfavfrt)
Myfavfrt.append("kiwi")
print(Myfavfrt)

# Q3 remove the 4th colour
colors = ["red", "green", "blue", "yellow", "orange"]
colors.pop(3)
print(colors)
```

```python
# convert a value to string and show its type
def myCode(value):
    tostr = str(value)
    print(tostr)
    print(type(tostr))
    return tostr

print(myCode(12345))
```

```python
# sum of positive and sum of negative numbers separately
num_list = [2, -3, 4, -1, -2, 1, 5, 3, -6, -4, 0, 7, -5]
sum_positive = sum(x for x in num_list if x > 0)
sum_negative = sum(x for x in num_list if x < 0)
print(sum_positive, sum_negative)
```

```python
# Do any two time intervals overlap?
pair_list = [(1, 5), (8, 9), (5, 6)]
overlap = any(
    max(a[0], b[0]) < min(a[1], b[1])
    for i, a in enumerate(pair_list)
    for b in pair_list[i + 1:]
)
print(overlap)   # False
```

```python
# even / odd split
l1 = [1, 2, 3, 4, 5, 6]
even = [i for i in l1 if i % 2 == 0]
odd = [i for i in l1 if i % 2 != 0]
print(even, odd)

# sum of only numeric items in a mixed list
mixed = [1, "a", 2.4, "b"]
print(sum(i for i in mixed if isinstance(i, (int, float))))
```
