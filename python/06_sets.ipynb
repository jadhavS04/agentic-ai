# 06 · Sets

**Topics covered:** properties · adding · removing · set operations · subset/superset · `frozenset` · set comprehension

## 1. Properties
- **Unique** elements only (no duplicates)
- **Unordered** – no index
- Written with `{}`
- **Mutable** – you can add/remove, but elements themselves must be immutable

```python
set1 = {12, 23, 34, 45, 56, 12}   # duplicate 12 is dropped
print(set1)

empty_set = set()                 # {} would create an empty DICT
print(type(empty_set), type({}))
```

## 2. Adding & removing

```python
set1 = {12, 23, 34, 45, 56}
set1.add(90)              # add one element
print(set1)
set1.update([67, 78])     # add many elements
print(set1)
set1.pop()                # removes a random element
print(set1)
set1.discard(57)          # remove, NO error if missing
set1.discard(23)
print(set1)

try:
    set1.remove(1000)     # remove raises KeyError if missing
except KeyError as e:
    print("KeyError:", e)
```

## 3. Set operations

```python
s = {1, 2, 3, 4}
s2 = {3, 4, 5, 6}
print(s | s2, s.union(s2))                         # union
print(s & s2, s.intersection(s2))                  # intersection
print(s - s2, s.difference(s2))                    # in s but not in s2
print(s2 - s, s2.difference(s))
print(s ^ s2, s.symmetric_difference(s2))          # in either, not both
```

## 4. Comparing sets

```python
s = {1, 2, 3, 4}
s3 = {1, 2}
print(s3.issubset(s))          # all of s3 is inside s
print(s.issuperset(s3))        # s contains all of s3
print(s.isdisjoint({10, 11}))  # no common elements?

s1 = s.copy()
print(s1)
s1.clear()
print(s1)
```

## 5. `frozenset` – an immutable set

```python
a = frozenset([1, 2, 3])
print(type(a), a)
try:
    a.add(4)
except AttributeError as e:
    print("Error:", e)
```

## 6. Set comprehension

```python
sq = {x * x for x in range(1, 6)}
print(sq)

# remove duplicates from a list
nums = [1, 2, 2, 3, 3, 3]
print(list(set(nums)))
```
