**Topics covered:** `try / except / else / finally` · multiple excepts · `Exception as e` · `raise` · retry loop · custom exceptions · file + exception

## 1. Why?
Errors at run-time crash the program. Exception handling lets us catch them and continue gracefully.

Common exceptions: `ZeroDivisionError`, `ValueError`, `TypeError`, `IndexError`, `KeyError`, `FileNotFoundError`, `NameError`

```python
try:
    a = int(input("Enter a: "))
    b = int(input("Enter b: "))
    print(a / b)
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
except ValueError:
    print("Error: Invalid input. Please enter valid integers.")
```

```python
# same code, invalid input
try:
    a = int(input("Enter a: "))
    b = int(input("Enter b: "))
    print(a / b)
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
except ValueError:
    print("Error: Invalid input. Please enter valid integers.")
```

### Bare `except:` catches everything – avoid it
It also hides unrelated bugs (and even the message you print may be wrong, e.g. a ValueError reported as "division by zero").

```python
try:
    a = int(input("Enter a: "))
    b = int(input("Enter b: "))
    print(a / b)
except:
    print("Something went wrong")
```

## 2. `else` and `finally`
- `else` → runs only if **no** exception happened
- `finally` → runs **always** (cleanup)

```python
try:
    a = int(input("Enter a: "))
    b = int(input("Enter b: "))
    print(a / b)
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
except ValueError:
    print("Error: Invalid input. Please enter valid integers.")
else:
    print("Division performed successfully.")
finally:
    print("Execution completed.")
```

## 3. `except Exception as e` – get the error message

```python
try:
    a = int(input("Enter a: "))
    b = int(input("Enter b: "))
    print(a / b)
except Exception as e:
    print(type(e).__name__, "->", e)
else:
    print("Division performed successfully.")
finally:
    print("Execution completed.")
```

## 4. Files + exceptions

```python
try:
    file = open("student.txt", "w")
    file.write("Hello,World!\nWELCOME TO PYTHON")
    file.close()

    file = open("student.txt", "r")
    print(file.read())
    file.close()
except Exception as e:
    print(e)
```

```python
# FileNotFoundError – with open() closes automatically, so no file.close() needed
try:
    with open("student1.txt", "r") as file:
        print(file.read())
except FileNotFoundError as e:
    print("Error:", e)
else:
    print("no error")
finally:
    print("finally block")

# (your original called file.close() in finally on a file that may never have opened –
#  that itself can raise NameError. Using `with` avoids this problem.)
```

```python
# seek() + FileNotFoundError handling (your "logging" example, with the \n typo fixed)
file = None
try:
    file = open("abc.txt", "w+")
    file.write("Welcome to exception handling\nwelcome to exception logging ")
    file.close()
    file = open("abc.txt", "r")
    file.seek(15)
    print(file.read())
except FileNotFoundError:
    print("File not found")
finally:
    if file:
        file.close()
```

## 5. `raise` – throw your own exception

```python
try:
    roll = int(input("enter your roll number: "))
    if roll <= 0:
        raise ValueError("Roll number must be positive")
    print("roll no:", roll)
except ValueError as e:
    print("ValueError Exception Thrown:", e)
```

```python
try:
    age = int(input("enter your age: "))
    if age < 18:
        raise Exception("Age must be 18 or older.")
    print("Welcome!")
except Exception as e:
    print(e)
```

## 6. Keep asking until the input is valid

```python
while True:
    try:
        num = int(input("Enter Number: "))
        print("Entered Number is:", num)
        break
    except ValueError:
        print("Invalid input. Please enter a valid integer.")
```

## 7. Custom exception class

```python
class MyCustomError(Exception):
    pass

class NegativeAgeError(Exception):
    def __init__(self, age, message="Age cannot be negative"):
        self.age = age
        self.message = message
        super().__init__(f"{message}: {age}")

roll = int(input("enter roll number: "))
try:
    if roll <= 0:
        raise MyCustomError("Roll number must be a positive integer.")
except MyCustomError as e:
    print("Caught:", e)

try:
    raise NegativeAgeError(-4)
except NegativeAgeError as e:
    print("Caught:", e)
```

## 8. Exception hierarchy tip
Put **specific** exceptions before general ones: `except ZeroDivisionError` → `except ValueError` → `except Exception`.
