## 1. Concepts
- **Class** – blueprint / template
- **Object** – instance of a class
- **Attribute** – data (variables) inside a class
- **Method** – function inside a class
- **`self`** – reference to the current object (always the first parameter of an instance method)

## 2. Class with class variables (attributes)

```python
class student:
    name = "nm"
    age = 23
    roll = 202

s = student()          # NOTE the () -> creates an object
print(s.name)
print(s.age)
print(s.roll)

# your original wrote `s = student` (no brackets) – that only gives another name
# for the CLASS, not an object. Both work for class variables, but s = student()
# is the correct way to create an object.
```

```python
class car:
    model = "nwxo"
    color = "white"
    price = 100000000

c = car()
print(c.model)
print(c.color)
print(c.price)

class car2:
    brand = "mybrand"
    model = "my model"

c1 = car2()
c2 = car2()
print(c1.brand)
print(c2.model)
```

## 3. Methods and `self`

```python
class student:
    def hefshine(self):
        print("hello")

s1 = student()
s2 = student()
s1.hefshine()
s2.hefshine()
```

```python
class Student:
    def get(self):
        self.name = input("enter name: ")
        self.address = input("enter address: ")

    def show(self):
        print("Name =", self.name)
        print("Address =", self.address)

stud = Student()
stud.get()
stud.show()
```

```python
class book:
    def mybook(self):
        self.title = input("enter title: ")
        self.author = input("enter author: ")

    def show(self):
        print("title =", self.title)
        print("author =", self.author)

b1 = book()
b1.mybook()
b1.show()
```

## 4. Passing values through method parameters

```python
class car:
    def model(self, name, color):
        self.name = name
        self.color = color

    def display(self):
        print("car name =", self.name)
        print("car color =", self.color)

c = car()
c.model("Toyota", "Red")
c.display()
```

```python
# area of rectangle + sum method
class rectangle:
    def area(self):
        self.length = int(input("enter length: "))
        self.breadth = int(input("enter breadth: "))
        area = self.length * self.breadth
        print("area =", area)

    def sum(self, a, b):
        self.a = a
        self.b = b
        return self.a + self.b

r = rectangle()
r.area()
print(r.sum(2, 3))
```

## 5. The constructor `__init__`
Runs **automatically** when an object is created – the best place to initialise attributes.
(Your earlier programs used `get()` / `model()` to set values; a constructor does the same job automatically.)

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def show(self):
        print(f"{self.name} earns {self.salary}")

e1 = Employee("shubham", 4000000)
e2 = Employee("rahul", 500000)
e1.show()
e2.show()
```

```python
# your Employee example – class body with input() runs ONCE when the class is defined
# and the values become shared class variables. Better to use __init__:
class Employee2:
    def __init__(self):
        self.name = input("enter name = ")
        self.salary = int(input("Enter salary = "))

e = Employee2()
print(e.name)
print(e.salary)
```

## 6. Instance variables vs class variables

```python
class Student:
    college = "LSPGCOER"            # class variable – shared by all objects

    def __init__(self, name, roll):
        self.name = name            # instance variable – separate for each object
        self.roll = roll

s1 = Student("a", 1)
s2 = Student("b", 2)
print(s1.college, s2.college)
Student.college = "XYZ"             # changing it on the class affects all objects
print(s1.college, s2.college)
print(s1.name, s2.name)
```

## 7. `__str__` – readable printing of an object

```python
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __str__(self):
        return f"Point({self.x}, {self.y})"

p = Point(3, 4)
print(p)
```

## 8. Practice programs
1. Create a `BankAccount` class with `deposit`, `withdraw`, `show_balance`.
2. Create a `Circle` class that computes area.

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
        else:
            print("Insufficient balance")

    def show_balance(self):
        print(self.owner, "balance:", self.balance)

acc = BankAccount("Shubham", 1000)
acc.deposit(500)
acc.withdraw(200)
acc.withdraw(5000)
acc.show_balance()
```
