# 09 · `while` Loop

**Topics covered:** 3 parts of a while loop · counting · odd/even · sum · reverse counting · list traversal · factorial · `break` / `continue` / `else`

## 1. Structure
1. **Initialization** – starting value
2. **Condition** – loop runs while this is `True`
3. **Increment / decrement** – change the value, otherwise → infinite loop

```python
# print numbers 1 to 10
i = 1                 # initialization
while i <= 10:        # condition
    print(i)          # body
    i += 1            # increment
```

```python
# odd numbers up to 20
num = 1
while num <= 20:
    print(num, end=" ")
    num += 2
```

```python
# even numbers between 50 and 70
i = 50
while i <= 70:
    if i % 2 == 0:
        print(i, end=" ")
    i = i + 1
```

```python
# sum of first 10 numbers
total = 0
i = 1
while i <= 10:
    total += i
    i += 1
print(total)
```

```python
# reverse counting
i = 10
while i >= 1:
    print(i, end=" ")
    i -= 1
```

```python
# traverse a list by index
Names = ["a", "b", "c", "d", "e"]
i = 0
while i < len(Names):
    print(Names[i])
    i += 1
```

## Factorial using while

```python
N = int(input("Enter num: "))
i = 1
fact = 1
while i <= N:
    fact *= i
    i += 1
print(f"factorial of {N} is {fact}")
```

## `break`, `continue`, `else`

```python
# break: stop the loop immediately
i = 1
while i <= 10:
    if i == 5:
        break
    print(i, end=" ")
    i += 1
print()

# continue: skip the rest of this iteration
i = 0
while i < 10:
    i += 1
    if i % 2 == 0:
        continue
    print(i, end=" ")
print()

# else runs if the loop ends WITHOUT break
i = 1
while i < 4:
    i += 1
else:
    print("loop finished normally")
```

## Problems on while loop

```python
# count digits of a number
num = int(input("enter num: "))
temp = abs(num)
c = 0
if temp == 0:
    c = 1
else:
    while temp > 0:
        temp = temp // 10
        c += 1
print(c)
```

```python
# reverse a number (handles negatives)
num = int(input("enter num: "))
temp = abs(num)
r_num = 0
while temp > 0:
    last_digit = temp % 10
    r_num = (r_num * 10) + last_digit
    temp = temp // 10
if num < 0:
    r_num = -r_num
print(r_num)
```

```python
# sum of digits
num = int(input("enter num: "))
temp = abs(num)
c = 0
while temp > 0:
    c += temp % 10
    temp = temp // 10
print(c)
```

```python
# power: base ** exponent without using **
b = float(input("Enter base: "))
p = int(input("Enter power: "))
r = 1
c = 0
while c < p:
    r = r * b
    c = c + 1
print(r)
```

```python
# multiplication table of 5
c = 1
while c <= 10:
    print(5, "x", c, "=", c * 5)
    c += 1
```

```python
# Fibonacci series up to N
N = int(input("enter num: "))
a = 0
b = 1
while a <= N:
    print(a, end=" ")
    a, b = b, a + b
```

```python
# prime number check
num = int(input("Enter a number: "))
if num <= 1:
    print(f"{num} is NOT a prime number.")
else:
    divisor = 2
    is_prime = True
    while divisor < num:
        if num % divisor == 0:
            is_prime = False
            break
        divisor += 1
    if is_prime:
        print(f"{num} is a prime number.")
    else:
        print(f"{num} is NOT a prime number.")
```
