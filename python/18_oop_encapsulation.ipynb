# 18 · OOP – Encapsulation (Data Hiding)

Bundle **data + methods** in one class and **restrict direct access** to the data.

**Topics covered:** public · protected (`_x`) · private (`__x`) · getters & setters · name mangling · `@property` link

## 1. Access levels (by convention)

| Syntax | Level | Meaning |
|---|---|---|
| `name` | public | accessible anywhere |
| `_name` | protected | "please don't touch" – meant for the class and its children |
| `__name` | private | name-mangled → `_ClassName__name` |

```python
class Student:
    def __init__(self):
        self.name = "Shubham"       # public
        self._roll = 20             # protected
        self.__marks = 90           # private

s = Student()
print(s.name)
print(s._roll)                      # works, but by convention you shouldn't
try:
    print(s.__marks)
except AttributeError as e:
    print("AttributeError:", e)

print(s._Student__marks)            # name mangling – still reachable, but discouraged
```

## 2. Getter & setter methods with validation

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
        else:
            print("Invalid amount")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient balance / invalid amount")

acc = BankAccount("Shubham", 5000)
acc.deposit(1000)
acc.withdraw(500)
acc.withdraw(99999)
print(acc.get_balance())

acc.balance = 0                      # creates a NEW public attribute; real balance is safe
print(acc.get_balance())
```

## 3. Private methods

```python
class ATM:
    def __init__(self, pin):
        self.__pin = pin

    def __verify(self, pin):         # private helper
        return pin == self.__pin

    def withdraw(self, pin, amount):
        if self.__verify(pin):
            print(f"{amount} withdrawn")
        else:
            print("Wrong PIN")

a = ATM(1234)
a.withdraw(1234, 2000)
a.withdraw(1111, 2000)
```

## 4. Encapsulation with `@property` (Pythonic way)
See notebook **21 · Descriptors** for the full explanation of `property`.

```python
class Person:
    def __init__(self, age):
        self.age = age               # goes through the setter

    @property
    def age(self):
        return self.__age

    @age.setter
    def age(self, value):
        if value < 0:
            raise ValueError("age cannot be negative")
        self.__age = value

p = Person(25)
print(p.age)
p.age = 30
print(p.age)
try:
    p.age = -5
except ValueError as e:
    print("ValueError:", e)
```
