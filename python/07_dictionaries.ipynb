# 07 · Dictionaries

**Topics covered:** properties · creating · accessing · adding/updating · removing · methods · nested dictionaries · dict comprehension · merging · practice programs

## 1. Properties
- Stores data as **key : value** pairs
- **Ordered** (insertion order, Python 3.7+), **changeable**
- **Keys must be unique** and immutable; values can be any type

```python
dict1 = {"k1": "v1", "k2": "v2"}
print(dict1, len(dict1), type(dict1))

# other ways to create
d1 = dict(name="shubham", age=22, course="python")
d2 = dict([("name", "shubham"), ("age", 22), ("course", "python")])
print(d1)
print(d2)
```

## 2. Accessing

```python
dict1 = {"k1": "v1", "k2": "v2"}
print(dict1["k1"])
print(dict1.get("k1"))
print(dict1.get("k4", "not found"))   # default value if key missing

try:
    dict1["k4"]                        # missing key -> KeyError
except KeyError as e:
    print("KeyError:", e)

print(dict1.keys())
print(dict1.values())
print(dict1.items())
```

```python
Book = dict(
    Name="MyBook",
    Price=1000,
    Purchase=["Flipkart", "Amazon"],
    Review=4.5,
    Auther="Alex"
)
print(Book["Name"], Book["Price"], Book.get("Price"))
print(Book)
```

## 3. Adding & updating

```python
Book["Publication"] = "kindley"                      # add new key
print(Book)
Book.update({"topics": ["topic1", "topic2"], "readtime": "23hr"})   # add several
print(Book)
Book["Name"] = "Book1"                               # change existing value
Book.update(Name="MyBook", Price=5000)               # update with keywords
print(Book)
```

## 4. Removing items

```python
car = {"nm": "car1", "mdl": 2016, "prc": 780000, "yr": 2020, "color": "red"}
car.pop("nm")         # remove by key (returns the value)
print(car)
car.popitem()         # remove the LAST inserted pair
print(car)
del car["mdl"]        # delete by key
print(car)
car.clear()           # empty the dict
print(car)
```

## 5. Copying, `fromkeys`, `setdefault`

```python
car = {"nm": "car1", "mdl": 2016, "prc": 780000}
new_car = car.copy()
new_dict = dict(car)
print(new_car, new_dict)

print(dict.fromkeys(new_dict, "Unknown"))   # same value for all keys
print(dict.fromkeys(new_dict))              # values are None

# setdefault: return existing value, or insert default if key missing
print(new_dict.setdefault("mdl", 2020))     # key exists -> returns 2016
print(new_dict.setdefault("grade", "A"))    # key missing -> inserts & returns
print(new_dict)
```

## 6. Nested dictionaries

```python
myfam = {
    "chld1": {"nm": "a", "age": 22, "yr": "4th"},
    "chld2": {"nm": "b", "age": 20, "yr": "1st",
              "marks": {"sub": 89, "sub1": 92, "sub2": 97}}
}
print(myfam)
print(myfam["chld1"])
print(myfam["chld1"]["nm"])
print(myfam["chld2"]["marks"]["sub1"])

myfam["chld1"]["nm"] = "g"       # update inner value
print(myfam["chld1"]["nm"])

del myfam["chld1"]               # delete a whole inner dict
print(myfam)
```

## 7. Dictionary comprehension

```python
n = [1, 2, 3, 4, 5]
n_cube = {num: num ** 3 for num in n if num % 2 != 0}
print(n_cube)
```

## 8. Looping over a dictionary

```python
marks = {"Alice": 85, "Bob": 72, "Charlie": 90}
for k in marks:
    print(k)
for v in marks.values():
    print(v)
for k, v in marks.items():
    print(k, "->", v)
print("Total:", sum(marks.values()))
```

## 9. Practice programs

```python
# find a key and return its value
def find_key(dict1, key):
    if key in dict1:
        return dict1[key]
    else:
        return "not found"

d = {1: 2, 3: 4, 5: 6}
print(find_key(d, 1))
print(find_key(d, 9))
```

```python
# invert a dictionary (swap keys and values)
fruit_prices = {"apple": 50, "banana": 20, "orange": 30, "mango": 60, "grapes": 40}
inverted = {value: key for key, value in fruit_prices.items()}
print(inverted)
```

```python
# merge two dictionaries (3 ways)
dict1 = {"name": "John", "age": 25}
dict2 = {"city": "New York", "country": "USA"}

print({**dict1, **dict2})     # unpacking
print(dict1 | dict2)          # union operator (3.9+)
merged = dict1.copy()
merged.update(dict2)          # update()
print(merged)
```

```python
# merge with a dict of dicts (later keys win)
student_marks = {"Alice": 85, "Bob": 72, "Charlie": 90, "David": 65, "Eva": 78}
students = {
    "Alice": {"Math": 85, "Science": 90},
    "Bob": {"Math": 70, "Science": 80},
    "Charlie": {"Math": 95, "Science": 88},
}
print(student_marks | students)
```
