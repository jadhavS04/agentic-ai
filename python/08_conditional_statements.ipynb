# 08 · Conditional Statements (`if / elif / else`)

**Topics covered:** `if` · `if-else` · `elif` ladder · nested `if` · ternary operator · ATM menu program

## 1. `if`, `if-else`, `elif`

```python
age = 20
if age >= 18:
    print("Eligible to vote")
else:
    print("Not eligible")

marks = 72
if marks >= 90:
    grade = "A+"
elif marks >= 75:
    grade = "A"
elif marks >= 60:
    grade = "B"
elif marks >= 40:
    grade = "C"
else:
    grade = "Fail"
print("Grade:", grade)
```

## 2. Nested `if`

```python
num = 15
if num > 0:
    if num % 2 == 0:
        print("positive even")
    else:
        print("positive odd")
else:
    print("zero or negative")
```

## 3. Ternary (one-line) `if`

```python
a, b = 10, 20
print("a is bigger" if a > b else "b is bigger")
```

## 4. Mini project – ATM menu
Uses nested conditions, `input()` and a PIN check.

```python
balance = 20000
pin = 1234

pin2 = int(input("Please enter your PIN: "))
if pin == pin2:
    print("1.deposit")
    print("2.withdraw")
    print("3.check balance")
    print("4.exit")
    ch = int(input("Enter your choice: "))
    if ch == 1:
        amt = int(input("Enter amount to deposit: "))
        if amt > 0:
            balance += amt
            print("Amount deposited successfully. New balance:", balance)
        else:
            print("Invalid amount. Please enter a positive value.")
    elif ch == 2:
        amt = int(input("Enter amount to withdraw: "))
        if amt <= balance:
            balance -= amt
            print("Amount withdrawn successfully. New balance:", balance)
        else:
            print("Insufficient balance.")
    elif ch == 3:
        print("Your current balance is:", balance)
    elif ch == 4:
        print("Thank you for using our service.")
    else:
        print("Invalid choice.")
else:
    print("Wrong PIN")
```
