## 1. Built-in data types (overview)

| Category | Types |
|---|---|
| Numeric | `int`, `float`, `complex` |
| Boolean | `bool` (`True` / `False`) |
| Sequence | `str`, `list`, `tuple` |
| Set | `set`, `frozenset` |
| Mapping | `dict` |
| None | `NoneType` |

```python
x = 10          # int
y = 3.14        # float
z = 2 + 3j      # complex
flag = True     # bool
name = "Shubham"            # str
marks = [90, 80, 70]        # list
point = (10, 20)            # tuple
unique = {1, 2, 3}          # set
student = {"name": "sj"}    # dict
nothing = None              # NoneType

for v in (x, y, z, flag, name, marks, point, unique, student, nothing):
    print(v, "->", type(v))
```

## 2. Variables
A variable is a **name that refers to a value** stored in memory. Syntax: `variable_name = value`

```python
name = "Shubham"
age = "22"          # note: this is a string because of the quotes
salary = 300000
is_student = True

# Student Name -> name, Roll Number -> roll_no, Marks -> marks
name = "Shubham"
roll_no = 20
marks = 90
print(name, roll_no, marks)
```

### Multiple assignment & assigning the same value

```python
# multiple variables, multiple values
x, y, z = 10, 20, 30
print(x)
print(y, z)

# same value to many variables
a = b = c = 10
print(a)
print(b, c)
```

### Swapping two variables
Python can swap without a temporary variable (tuple packing/unpacking).

```python
a = 10
b = 20
a, b = b, a
print(a, b)   # 20 10
```

## 3. Dynamic typing
Python **detects the data type automatically** from the value; the same name can later hold a different type.

```python
x, y, z = 10, "python", 3.14
print(type(x), type(y), type(z))

# type() takes ONE argument -> type(x, y, z) would be an error
v = 5
print(type(v))
v = "now a string"
print(type(v))
```

```python
name = "Shubham"
age = 22
branch = "Artificial Intelligence And Data Science"
print(name, age, branch)
```

## 4. Rules for naming variables

1. Must start with a **letter or underscore** (not a digit)
2. Can contain **letters, digits and underscore** only
3. **No spaces** and no special characters (`@`, `-`, `$` ...)
4. Names are **case sensitive** (`age` ≠ `Age`)
5. Cannot be a Python **keyword**
6. Use **meaningful** names

```python
name = "sj"
_age = 20
print(name, _age)

marks10 = 90          # digits allowed, but not at the start
print(marks10)

Stud_nm = "mY"
r_n_1 = 101
print(Stud_nm)        # case sensitive: print(stud_nm) would give NameError
print(r_n_1)

# INVALID names (uncomment to see the SyntaxError):
# 1name = "sj"        # starts with a digit
# 10marks = 90
# student-nm = "alex" # hyphen not allowed
# stud@nm = "alex"    # special char
# stud nm = "ab"      # space

# case sensitivity
age = 20
Age = 30
print(age, Age)       # two different variables

# Names that merely LOOK like keywords are fine because of different case/spelling
Class_nm = "py"
for_cont = 1

# bad vs good naming
a = 100               # unclear
b = "ramesh"          # unclear
salary = 200000       # good
Stud_nm = "rahul"     # good
```

## 6. Escape characters
A backslash `\` inside a string gives a character a special meaning.

| Escape | Meaning |
|---|---|
| `\'` `\"` | quote inside a string |
| `\\` | backslash |
| `\n` | new line |
| `\t` | tab |
| `\r` | carriage return |
| `\b` | backspace |
| `\f` | form feed |
| `\ooo` | octal value |
| `\xhh` | hex value |

```python
print('It\'s a python class.')
print('C:\\Python\\Files')
print("hello\nworld")
print("c1\tc2")
print('Octal value: \142')   # \142 -> 'b'
print('h\rw')                 # carriage return
print('hello\b')              # backspace
print('hello\f World')        # form feed
print('\x62')                 # hex -> 'b'

# raw string: ignore escapes
print(r"C:\new\table")
```

## 8. Type casting (conversion)

```python
print(int("25") + 5)       # str  -> int
print(float(7))            # int  -> float
print(str(123) + "abc")    # int  -> str
print(bool(0), bool(5), bool(""), bool("a"))
print(list("abc"), tuple([1, 2]), set([1, 1, 2]))

# int() on a non-number raises ValueError
try:
    int("abc")
except ValueError as e:
    print("Error:", e)
```
